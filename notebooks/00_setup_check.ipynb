# Setup check

Run all cells. If this executes without errors and prints versions below, the environment and kernel are wired up correctly.

```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import sklearn

print("Python:", sys.version)
print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("statsmodels:", sm.__version__)
print("scikit-learn:", sklearn.__version__)
```

```python
# Quick plot + simple OLS regression, just to confirm matplotlib and statsmodels both work end to end
rng = np.random.default_rng(0)
x = np.linspace(0, 10, 50)
y = 2 * x + 1 + rng.normal(0, 1, size=50)

X = sm.add_constant(x)
model = sm.OLS(y, X).fit()
print(model.params)

plt.scatter(x, y, label="data")
plt.plot(x, model.predict(X), color="red", label="OLS fit")
plt.legend()
plt.title("Setup check: OLS on synthetic data")
plt.show()
```
