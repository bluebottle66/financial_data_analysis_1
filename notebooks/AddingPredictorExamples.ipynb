```python
import statsmodels.api as sm
from scipy.stats import norm
import pandas as pd
import matplotlib.pyplot as plt
```

```python
n = 100
```

## Example One

```python
df1 = pd.DataFrame()

df1['X1'] = norm.rvs(size=n,random_state=0)
df1['X2'] = df1['X1'] + norm.rvs(scale=0.4,size=n,random_state=1)
df1['Y'] = 10 + 0.2*df1['X1'] + norm.rvs(size=n,random_state=2)
```

```python
sm.OLS.from_formula("Y ~ X1", data=df1).fit().summary()
```

```python
sm.OLS.from_formula("Y ~ X1 + X2", data=df1).fit().summary()
```

```python
fig,axs = plt.subplots(1,3,figsize=[15,5])

axs[0].plot(df1['X1'],df1['Y'],marker=".",linestyle="none",ms=5)
axs[0].set_xlabel("X1",size=14)
axs[0].set_ylabel("Y",size=14)

axs[1].plot(df1['X2'],df1['Y'],marker=".",linestyle="none",ms=5)
axs[1].set_xlabel("X2",size=14)
axs[1].set_ylabel("Y",size=14)

axs[2].plot(df1['X1'],df1['X2'],marker=".",linestyle="none",ms=5)
axs[2].set_xlabel("X1",size=14)
axs[2].set_ylabel("X2",size=14)

plt.show()
```

## Example Two

```python
df2 = pd.DataFrame()

df2['X1'] = norm.rvs(size=n,random_state=0)
df2['X2'] = norm.rvs(size=n,random_state=3)
df2['Y'] = 10 + 0.2*df2['X1'] + 2*df2['X2'] + norm.rvs(size=n,random_state=2)
```

```python
sm.OLS.from_formula("Y ~ X1", data=df2).fit().summary()
```

```python
sm.OLS.from_formula("Y ~ X1 + X2", data=df2).fit().summary()
```

```python
fig,axs = plt.subplots(1,3,figsize=[15,5])

axs[0].plot(df2['X1'],df2['Y'],marker=".",linestyle="none",ms=5)
axs[0].set_xlabel("X1",size=14)
axs[0].set_ylabel("Y",size=14)

axs[1].plot(df2['X2'],df2['Y'],marker=".",linestyle="none",ms=5)
axs[1].set_xlabel("X2",size=14)
axs[1].set_ylabel("Y",size=14)

axs[2].plot(df2['X1'],df2['X2'],marker=".",linestyle="none",ms=5)
axs[2].set_xlabel("X1",size=14)
axs[2].set_ylabel("X2",size=14)

plt.show()
```

## Example Three

```python
n=1000
```

```python
df3 = pd.DataFrame()

df3['X1'] = norm.rvs(size=n,random_state=0)
df3['X2'] = df3['X1'] + norm.rvs(size=n,random_state=3)

df3['Y'] = 10 + 1*df3['X1'] - 2*df3['X2'] + norm.rvs(size=n,random_state=2,scale=0.1)
```

```python
sm.OLS.from_formula("Y ~ X1", data=df3).fit().summary()
```

```python
sm.OLS.from_formula("Y ~ X1 + X2", data=df3).fit().summary()
```

```python
fig,axs = plt.subplots(1,3,figsize=[15,5])

axs[0].plot(df3['X1'],df3['Y'],marker=".",linestyle="none",ms=5)
axs[0].set_xlabel("X1",size=14)
axs[0].set_ylabel("Y",size=14)

axs[1].plot(df3['X2'],df3['Y'],marker=".",linestyle="none",ms=5)
axs[1].set_xlabel("X2",size=14)
axs[1].set_ylabel("Y",size=14)

axs[2].plot(df3['X1'],df3['X2'],marker=".",linestyle="none",ms=5)
axs[2].set_xlabel("X1",size=14)
axs[2].set_ylabel("X2",size=14)

plt.show()
```

```python
eps = 0.1

fig,axs = plt.subplots(1,3,figsize=[15,5])

df3b = df3[abs(df3['X2']-1)<eps]
axs[0].plot(df3b['X1'],df3b['Y'],marker=".",linestyle="none",ms=5)
axs[0].set_xlabel("X1",size=14)
axs[0].set_ylabel("Y",size=14)
axs[0].set_title("Slice when X2 = -1")

df3b = df3[abs(df3['X2']-0)<eps]
axs[1].plot(df3b['X1'],df3b['Y'],marker=".",linestyle="none",ms=5)
axs[1].set_xlabel("X1",size=14)
axs[1].set_ylabel("Y",size=14)
axs[1].set_title("Slice when X2 = 0")

df3b = df3[abs(df3['X2']+1)<eps]
axs[2].plot(df3b['X1'],df3b['Y'],marker=".",linestyle="none",ms=5)
axs[2].set_xlabel("X1",size=14)
axs[2].set_ylabel("Y",size=14)
axs[2].set_title("Slice when X2 = 1")

plt.show()
```

```python
fig,axs = plt.subplots(1,1,figsize=[4.8,5])

scat = axs.scatter(df3['X1'],df3['Y'],marker=".")
axs.set_xlabel("X1",size=14)
axs.set_ylabel("Y",size=14)
plt.show()
```

```python
fig,axs = plt.subplots(1,1,figsize=[6,5])

scat = axs.scatter(df3['X1'],df3['Y'],marker=".", c=df3['X2'],
                  cmap="coolwarm")
axs.set_xlabel("X1",size=14)
axs.set_ylabel("Y",size=14)

fig.colorbar(scat,label="X2")
plt.show()
```
