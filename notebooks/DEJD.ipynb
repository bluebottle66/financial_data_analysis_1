# Part 11: Case Study: The DEJD Model {-}

This Part will explore a **jump diffusion model**, namely the **double exponential jump diffusion (DEJD) model**.
This was described in Kou (2002) (*Management Science*, Volume 48), but here we will use go through a paper by Ramezani and Zeng (2007) (*Annals of Finance*, Volume 3). An annotated version of this paper is available on Canvas; references made below are to the margin numbers in that version.

Instead of using the normal distribution to model jump sizes (as in the **Merton model**), the DEJD model uses exponential distributions. A distinction is made between "up jumps" and "down jumps," i.e., each has its own parameter for its corresponding exponential distribution.

**Passage 1**

As is typical, the focus when creating the model is on having it not only reproduces features of real data, but also leads to (relatively) simple mathematical results. In this case, the authors mention that "DEJD leads to tractable pricing formulas for exotic and path dependent options."

A **leptokurtic distribution** is one whose kurtosis is larger than 3 (positive excess kurtosis). Such distributions have heavier tails than the normal distribution.

The expression **exotic option** doesn't really have a fixed definition, but refers to options which are not the basic **vanilla** call and put options. For example...

1. There are options on options, i.e., you can have a call option that gives you the right to purchase a call option at a certain price.
2. **Barrier options** are cancelled if the price of the underlying asset reaches a certain level.
3. With an **Asian option**, the payoff amount is calculated by taking the average of the price of underlying asset over some time period.

A **path dependent option** is one for which the payoff amount depends on the entire history of the underlying asset's price over the time the option is held.

**Passage 2**

Note the use of maximum likelihood for estimating the parameters. That is a key contribution of this work.

**Passage 3**

The authors start by describing the basic properties of the Merton model. Note that they state that the "jump magnitudes are log-normally distributed." In our discussion, we focused on the price on the log scale, and hence the jumps were normally distributed.

**Passage 4**

As usual, there are a number of studies from people working to improve/enhance the Merton model. Having "time varying jump intensity" would mean that the rate of the Poisson process could vary with time.

**Passage 5**

A key limitation of the Merton model: there is a single Poisson process describing the occurrence of jumps, and the jump size distribution (on log scale) is symmetric around zero.

\newpage

**Passage 6**

Kou's model also has a single Poisson process to model the occurrence of jumps, but when a jump occurs it is an up jump with probability $p$. Ramezani and Zeng (1998) use two independent Poisson processes to model the up jumps and down jumps. As we have seen, the combined process remains a Poisson process. These two modeling approaches are equivalent.

The difference between these authors is in the distribution assumed for the jump sizes. 
Kou assumes that **on the log scale**, the size of a jump has an exponential distribution. Jumps up have a mean size of $1/\eta_1$,
while jumps down have a mean size of $1/\eta_2$.
Ramezani and Zeng work **on the original price scale**, and assume jumps up are drawn from a Pareto distribution, while down jumps are drawn from a Beta distribution. These jumps **multiply** the current price, hence are additive on the log scale.

**Passage 9**

The distinguish the model which uses the Pareto and Beta distributions (the *PBJD model*) from the DEJD model. We will see, however, that they are equivalent in some cases.

**Passage 10**

Again, this model is constructing multiplicative factors to represent the jumps. 

The Beta distribution has support $(0,1)$, and prices cannot go negative, so mutiplying by a Beta random variable seems like a reasonable model for the down jumps.

The Pareto is chosen to have support $(1,\infty)$. The result is that multiplying by a random variable with this distribution will definitely increase the price.
