```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
```

```python
data = {"Engine_Size": [1.6, 2.0, 2.4, 3.0, 3.5], "HP": [105, 140, 165, 210, 250]}
df = pd.DataFrame(data)
```

```python
X = df["Engine_Size"]
y = df["HP"]
```

```python
print (X)
```

```python
X = sm.add_constant(X)
```

```python
print (X)
```

```python
print(y)
```

```python
model = sm.OLS(y, X).fit()
print(model.summary())
```
